```python
## III. Inspección del modelo S. liquefaciens UNJFSC 002
import cobra
Serratia = cobra.io.read_sbml_model('modelos/SERRATIA_polished.sbml')
```

```python
#Inspección del modelo Metabolico Serratia liquefaciens UNJFSC 002
Serratia
```

```python
#Inspeccionando el modelo
model_list = [Serratia]
for model in model_list:
    num_reactions = len(model.reactions)
    num_metabolites = len(model.metabolites)
    num_genes = len(model.genes)

    print("Model information")
    print("Model ID is: {}".format(model.id))
    print("Number of reactions: {}".format(num_reactions))
    print("Number of metabolites: {}".format(num_metabolites))
    print("Number of genes: {}".format(num_genes))
    print("Model compartments:")
    for k, v in model.compartments.items():
        print(k + '\t' + v)
    print('\n################################\n')
```

```python
#Inspeccionando el modelo
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
%matplotlib notebook 
%matplotlib inline

labels = ['Serratia_liquefaciens']
num_reactions = [len(Serratia.reactions)]
num_metabolites = [len(Serratia.metabolites)]
num_genes = [len(Serratia.genes)]

x = np.arange(len(labels))  # the label locations
width = 0.3  # the width of the bars

fig, ax = plt.subplots()
rects1 = ax.bar(x - width, num_reactions, width, label='# de Reacciones')
rects2 = ax.bar(x, num_metabolites, width, label='# de Metabolitos')
rects3 = ax.bar(x + width, num_genes, width, label='# de Genes')

# Add some text for labels, title and custom x-axis tick labels, etc.
ax.set_ylabel('Cantidad')
ax.set_title('Serratia liquefaciens UNJFSC002')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()


def autolabel(rects):
    """Attach a text label above each bar in *rects*, displaying its height."""
    for rect in rects:
        height = rect.get_height()
        ax.annotate('{}'.format(height),
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')


autolabel(rects1)
autolabel(rects2)
autolabel(rects3)
fig.set_size_inches(7.5, 5.5)
#fig.tight_layout()
plt.show()
```

```python
# codigos para Rx especificas identificadas por KEGG pero exportables en un formato como CSV
import pandas as pd

reacciones_anotadas = []

for reaccion in model.reactions:
    if 'kegg.reaction' in reaccion.annotation:
        kegg_id = reaccion.annotation['kegg.reaction']
        if isinstance(kegg_id, list):
            kegg_id = ', '.join(kegg_id)
        reacciones_anotadas.append({
            'ID de la Reacción': reaccion.id,
            'Nombre': reaccion.name,
            'Fórmula': reaccion.reaction,
            'Genes Asociados': reaccion.gene_reaction_rule,
            'ID de KEGG': kegg_id
        })

df_reacciones = pd.DataFrame(reacciones_anotadas)

df_reacciones.to_csv('reacciones_anotadas_kegg.csv', index=False)

print(f"Número total de reacciones en el modelo: {len(model.reactions)}")

print(f"Número de reacciones con anotación de KEGG: {len(df_reacciones)}")
```

```python
# Rx encontradas en Reactome, Metanetx y chebi
import pandas as pd

datos_reacciones_reactome = []
datos_reacciones_metanetx = []
datos_reacciones_chebi = []

for reaccion in model.reactions:
    # Verificar si la reacción tiene una anotación de Reactome
    if 'reactome' in reaccion.annotation:
        reactome_id = reaccion.annotation['reactome']
        # Si es una lista, unir los elementos en una cadena separada por comas
        if isinstance(reactome_id, list):
            reactome_id = ', '.join(reactome_id)
        datos_reacciones_reactome.append({
            'ID de la Reacción': reaccion.id,
            'Nombre': reaccion.name,
            'Fórmula': reaccion.reaction,
            'Genes Asociados': reaccion.gene_reaction_rule,
            'ID de Reactome': reactome_id
        })
    
    if 'metanetx.reaction' in reaccion.annotation:
        metanetx_id = reaccion.annotation['metanetx.reaction']
        if isinstance(metanetx_id, list):
            metanetx_id = ', '.join(metanetx_id)
        datos_reacciones_metanetx.append({
            'ID de la Reacción': reaccion.id,
            'Nombre': reaccion.name,
            'Fórmula': reaccion.reaction,
            'Genes Asociados': reaccion.gene_reaction_rule,
            'ID de MetaNetX': metanetx_id
        })
    
    if 'chebi' in reaccion.annotation:
        chebi_id = reaccion.annotation['chebi']
        if isinstance(chebi_id, list):
            chebi_id = ', '.join(chebi_id)
        datos_reacciones_chebi.append({
            'ID de la Reacción': reaccion.id,
            'Nombre': reaccion.name,
            'Fórmula': reaccion.reaction,
            'Genes Asociados': reaccion.gene_reaction_rule,
            'ID de ChEBI': chebi_id
        })

df_reacciones_reactome = pd.DataFrame(datos_reacciones_reactome)
df_reacciones_metanetx = pd.DataFrame(datos_reacciones_metanetx)
df_reacciones_chebi = pd.DataFrame(datos_reacciones_chebi)

df_reacciones_reactome.to_csv('reacciones_reactome_modelo.csv', index=False)
df_reacciones_metanetx.to_csv('reacciones_metanetx_modelo.csv', index=False)
df_reacciones_chebi.to_csv('reacciones_chebi_modelo.csv', index=False)

# Imprimir estadísticas del modelo
print(f"Número total de reacciones en el modelo: {len(model.reactions)}")
print(f"Número de reacciones con anotaciones de Reactome: {len(df_reacciones_reactome)}")
print(f"Número de reacciones con anotaciones de MetaNetX: {len(df_reacciones_metanetx)}")
print(f"Número de reacciones con anotaciones de ChEBI: {len(df_reacciones_chebi)}")
```
